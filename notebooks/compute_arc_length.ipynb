```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipeinc, ellipe

```

```python
# 1. Define parameters
a = 10.0
b = 1.0
phi_degrees = 80

# 2. Convert angle to radians
phi_radians = np.deg2rad(phi_degrees)

# 3. Calculate the parameter m (which is k^2) for SciPy functions
m = 1 - (b**2 / a**2)

# 4. Calculate the incomplete elliptic integral E(phi | m)
E_phi = ellipeinc(phi_radians, m)

# 5. Calculate the complete elliptic integral E(m) for the full quadrant
E_quadrant = ellipe(m)

# 6. Calculate the percentage of arc length
percentage = (E_phi / E_quadrant) * 100

# Create figure
fig, ax = plt.subplots(1, 1, figsize=(10, 8))

# Plot circumscribed circle (radius = semi-major axis)
theta_circle = np.linspace(0, 2*np.pi, 1000)
x_circle = a * np.cos(theta_circle)
y_circle = a * np.sin(theta_circle)
ax.plot(x_circle, y_circle, 'c--', linewidth=1.5, alpha=0.6, label='Circumscribed Circle')

# Plot full ellipse
theta_full = np.linspace(0, 2*np.pi, 1000)
x_full = a * np.cos(theta_full)
y_full = b * np.sin(theta_full)
ax.plot(x_full, y_full, 'b-', linewidth=2, label='Full Ellipse')

# Plot the highlighted arc (from 0 to phi_radians)
theta_arc = np.linspace(0, phi_radians, 100)
x_arc = a * np.cos(theta_arc)
y_arc = b * np.sin(theta_arc)
ax.plot(x_arc, y_arc, 'r-', linewidth=4, label=f'Arc for φ = {phi_degrees}°')

# Point on ellipse
x_point = a * np.cos(phi_radians)
y_point = b * np.sin(phi_radians)

# Point on circle (same parametric angle)
x_circle_point = a * np.cos(phi_radians)
y_circle_point = a * np.sin(phi_radians)

# Plot the parametric angle line from origin (extends to circle)
ax.plot([0, x_circle_point], [0, y_circle_point], 'g--', linewidth=2, label=f'Parametric Angle ({phi_degrees}°)')

# Draw vertical line showing projection from circle to ellipse
ax.plot([x_point, x_circle_point], [y_point, y_circle_point], 'm:', linewidth=1.5, alpha=0.7, label='Projection')

# Mark the point on the ellipse
ax.plot(x_point, y_point, 'ro', markersize=10, label=f'Point on Ellipse')

# Mark the point on the circle
ax.plot(x_circle_point, y_circle_point, 'co', markersize=10, label=f'Point on Circle')

# Mark the origin
ax.plot(0, 0, 'ko', markersize=8, label='Origin')

# Draw the semi-major and semi-minor axes
ax.plot([0, a], [0, 0], 'k-', linewidth=1, alpha=0.5)
ax.plot([0, 0], [0, b], 'k-', linewidth=1, alpha=0.5)
ax.text(a/2, -0.5, f'a = {a}', ha='center', fontsize=10)
ax.text(-0.5, b/2, f'b = {b}', ha='right', fontsize=10)

# Add annotations
ax.annotate(f'φ = {phi_degrees}°',
            xy=(x_circle_point/3, y_circle_point/3),
            xytext=(x_circle_point/3 + 2, y_circle_point/3 + 1),
            arrowprops=dict(arrowstyle='->', color='green', lw=1.5),
            fontsize=12, color='green')

ax.text(0.5, -3, f'Arc Length Percentage: {percentage:.2f}%',
        fontsize=14, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Set equal aspect ratio and labels
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title(f'Ellipse Arc Length Calculation with Parametric Angle\n(a={a}, b={b}, φ={phi_degrees}°)',
             fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=9)

# Set axis limits
ax.set_xlim(-a*1.2, a*1.2)
ax.set_ylim(-a*1.2, a*1.2)

plt.tight_layout()
plt.show()

# Print summary
print(f"\nEllipse Parameters:")
print(f"  Semi-major axis (a): {a}")
print(f"  Semi-minor axis (b): {b}")
print(f"  Parametric angle (φ): {phi_degrees}°")
print(f"\nArc Length Calculation:")
print(f"  Eccentricity squared (m): {m:.4f}")
print(f"  Incomplete E(φ | m): {E_phi:.4f}")
print(f"  Complete E(m) for quadrant: {E_quadrant:.4f}")
print(f"  Arc length percentage: {percentage:.2f}%")
print(f"\nParametric Angle Demonstration:")
print(f"  Point on circle: ({x_circle_point:.2f}, {y_circle_point:.2f})")
print(f"  Point on ellipse: ({x_point:.2f}, {y_point:.2f})")
print(f"  The parametric angle is measured on the circumscribed circle")
```
