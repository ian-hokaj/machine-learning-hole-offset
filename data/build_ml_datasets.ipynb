## Create train/test split where test set contains all values along a fixed parameter combination

```python
def create_custom_train_test_split(dataset, target_params, param_names=['w/r', 'a/c', 'a/t', 'r/t']):
    """
    Create train/test split based on matching any of the target parameters.
    
    Parameters:
    -----------
    dataset : np.ndarray
        Full dataset with shape (n_samples, 5) where columns are [w/r, a/c, a/t, r/t, Kbearing_c]
    target_params : list or np.ndarray
        Target parameter values [w/r, a/c, a/t, r/t]
    param_names : list
        Names of the parameters for display
    
    Returns:
    --------
    X_train, y_train, X_test, y_test : np.ndarray
        Training and test features and labels
    test_count : int
        Number of test samples
    """
    
    w_over_r_target, a_over_c_target, a_over_t_target, r_over_t_target = target_params
    
    train_data = []
    test_data = []
    
    # Iterate through entire dataset
    for row in dataset:
        w_over_r, a_over_c, a_over_t, r_over_t, kbearing_c = row
        
        # Check if any parameter matches the target
        if (np.isclose(w_over_r, w_over_r_target) or 
            np.isclose(a_over_c, a_over_c_target) or 
            np.isclose(a_over_t, a_over_t_target) or 
            np.isclose(r_over_t, r_over_t_target)):
            test_data.append(row)
        else:
            train_data.append(row)
    
    train_data = np.array(train_data)
    test_data = np.array(test_data)
    
    # Split features and labels
    X_train = train_data[:, :4]
    y_train = train_data[:, 4:5]
    X_test = test_data[:, :4]
    y_test = test_data[:, 4:5]
    
    # Print summary
    print("="*80)
    print("CUSTOM TRAIN/TEST SPLIT SUMMARY")
    print("="*80)
    print(f"\nTarget Parameters:")
    for i, (name, value) in enumerate(zip(param_names, target_params)):
        print(f"  {name}: {value}")
    
    print(f"\nDataset Split:")
    print(f"  Training samples: {len(X_train)}")
    print(f"  Test samples: {len(X_test)}")
    print(f"  Total samples: {len(dataset)}")
    print(f"  Test set percentage: {len(X_test)/len(dataset)*100:.2f}%")
    
    print(f"\nFeature Shapes:")
    print(f"  X_train: {X_train.shape}")
    print(f"  y_train: {y_train.shape}")
    print(f"  X_test: {X_test.shape}")
    print(f"  y_test: {y_test.shape}")
    
    print("="*80)
    
    return X_train, y_train, X_test, y_test


# Load the dataset
dataset_name = "four_params_Kbearing_c_outliersRemoved_finiteWidthFiltered15"
dataset = np.load(f'ml_formatted_datasets/{dataset_name}.npy')

# Define target parameter combination
# print(np.unique(dataset[:,0]))
# print(np.unique(dataset[:,1]))
# print(np.unique(dataset[:,2]))
# print(np.unique(dataset[:,3]))
target_params = [4.0, 1.0, 0.5, 1.0]  # [w/r, a/c, a/t, r/t] - modify as needed

# Create the train/test split
X_train, y_train, X_test, y_test = create_custom_train_test_split(
    dataset,
    target_params,
    param_names=['w/r', 'a/c', 'a/t', 'r/t']
)

# Optionally save the datasets
os.makedirs(f'ml_formatted_datasets/singleParamSlice_{dataset_name}', exist_ok=True)
np.save(f'ml_formatted_datasets/singleParamSlice_{dataset_name}/X_train.npy', X_train)
np.save(f'ml_formatted_datasets/singleParamSlice_{dataset_name}/y_train.npy', y_train)
np.save(f'ml_formatted_datasets/singleParamSlice_{dataset_name}/X_test.npy', X_test)
np.save(f'ml_formatted_datasets/singleParamSlice_{dataset_name}/y_test.npy', y_test)
```

```python
import os
import glob
import pandas as pd

# Get all directories in ml_formatted_datasets
dataset_dirs = [d for d in os.listdir('ml_formatted_datasets') 
                if os.path.isdir(os.path.join('ml_formatted_datasets', d))]

dataset_dirs.sort()

print("="*120)
print("ANALYSIS OF ALL ML FORMATTED DATASETS")
print("="*120)

# Store results for summary table
all_results = []

for dataset_dir in dataset_dirs:
    dataset_path = os.path.join('ml_formatted_datasets', dataset_dir)
    
    # Check if train/test files exist
    X_train_path = os.path.join(dataset_path, 'X_train.npy')
    y_train_path = os.path.join(dataset_path, 'y_train.npy')
    X_test_path = os.path.join(dataset_path, 'X_test.npy')
    y_test_path = os.path.join(dataset_path, 'y_test.npy')
    
    # Skip if files don't exist
    if not all([os.path.exists(p) for p in [X_train_path, y_train_path, X_test_path, y_test_path]]):
        continue
    
    # Load the data
    X_train = np.load(X_train_path)
    y_train = np.load(y_train_path)
    X_test = np.load(X_test_path)
    y_test = np.load(y_test_path)
    
    print(f"\n{'='*120}")
    print(f"DATASET: {dataset_dir}")
    print(f"{'='*120}")
    
    # Print basic statistics
    print(f"\nDATA SHAPE:")
    print(f"  X_train: {X_train.shape}")
    print(f"  y_train: {y_train.shape}")
    print(f"  X_test: {X_test.shape}")
    print(f"  y_test: {y_test.shape}")
    
    print(f"\nTRAINING SET STATISTICS:")
    print(f"  Number of samples: {len(X_train)}")
    print(f"  y_train - Mean: {np.mean(y_train):.4f}, Std Dev: {np.std(y_train):.4f}")
    print(f"  y_train - Min: {np.min(y_train):.4f}, Max: {np.max(y_train):.4f}")
    
    print(f"\nTEST SET STATISTICS:")
    print(f"  Number of samples: {len(X_test)}")
    print(f"  y_test - Mean: {np.mean(y_test):.4f}, Std Dev: {np.std(y_test):.4f}")
    print(f"  y_test - Min: {np.min(y_test):.4f}, Max: {np.max(y_test):.4f}")
    
    print(f"\nFEATURE STATISTICS (Training Set):")
    feature_names = ['w/r', 'a/c', 'a/t', 'r/t']
    for i, name in enumerate(feature_names):
        if i < X_train.shape[1]:
            print(f"  {name} - Mean: {np.mean(X_train[:, i]):.4f}, Std Dev: {np.std(X_train[:, i]):.4f}, " +
                  f"Min: {np.min(X_train[:, i]):.4f}, Max: {np.max(X_train[:, i]):.4f}")
    
    print(f"\nFEATURE STATISTICS (Test Set):")
    for i, name in enumerate(feature_names):
        if i < X_test.shape[1]:
            print(f"  {name} - Mean: {np.mean(X_test[:, i]):.4f}, Std Dev: {np.std(X_test[:, i]):.4f}, " +
                  f"Min: {np.min(X_test[:, i]):.4f}, Max: {np.max(X_test[:, i]):.4f}")
    
    # Store for summary
    all_results.append({
        'Dataset': dataset_dir,
        'Train Samples': len(X_train),
        'Test Samples': len(X_test),
        'Total Samples': len(X_train) + len(X_test),
        'Train y Mean': np.mean(y_train),
        'Train y Std': np.std(y_train),
        'Test y Mean': np.mean(y_test),
        'Test y Std': np.std(y_test),
        'Train/Total %': f"{len(X_train)/(len(X_train) + len(X_test))*100:.1f}%"
    })

print(f"\n{'='*120}")
print("SUMMARY TABLE")
print(f"{'='*120}\n")

# Create and display summary table
summary_df = pd.DataFrame(all_results)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.float_format', '{:.4f}'.format)

print(summary_df.to_string(index=False))
print(f"\n{'='*120}")
```
